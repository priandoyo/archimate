```python
from graphviz import Digraph
from IPython.display import display, Image

# Create diagram
dot = Digraph("TOGAF_DBRM", format="png")

dot.attr(
    rankdir="TB",
    bgcolor="white",
    pad="0.4",
    nodesep="0.4",
    ranksep="0.6",
    label="TOGAF Digital Business Reference Model by Anjar Priandoyo",
    labelloc="t",
    fontsize="22",
    fontname="Arial"
)

dot.attr(
    "node",
    shape="box",
    style="rounded,filled",
    fontname="Arial",
    fontsize="11",
    color="#64748B",
    margin="0.2"
)

domains = {
    "Digital Domain": {
        "color": "#DBEAFE",
        "elements": [
            "Customer Focus",
            "Digital Enterprise"
        ]
    },
    "Strategy Domain": {
        "color": "#FFEDD5",
        "elements": [
            "Strategic Context",
            "Business Motivation",
            "Business Services and Products"
        ]
    },
    "Structural Domain": {
        "color": "#DCFCE7",
        "elements": [
            "Ecosystem and Business Model",
            "Operating Model",
            "Business Capability Model"
        ]
    },
    "Operational Domain": {
        "color": "#EDE9FE",
        "elements": [
            "Business Operating Environment",
            "Digital Enablement"
        ]
    }
}

# Generate domain containers
for index, (domain, data) in enumerate(domains.items()):

    with dot.subgraph(name=f"cluster_{index}") as c:

        c.attr(
            label=domain,
            style="rounded,filled",
            color=data["color"],
            fillcolor=data["color"],
            fontsize="14",
            fontname="Arial",
            penwidth="2"
        )

        for i, element in enumerate(data["elements"]):
            node_id = f"d{index}_e{i}"

            c.node(
                node_id,
                label=element,
                fillcolor="white",
                fontcolor="#1E293B"
            )

        # Keep elements horizontally aligned
        c.attr(rank="same")

# Render directly in Google Colab
output = dot.pipe(format="png")
display(Image(output))
```
